The perceptron is called a linear classifier because it separates data using a linear boundary.

A dataset is linearly separable if we can draw a single straight line that separates data points of different classes.

# Question 04: [ Marks 15 ]

Write the perceptron weight update rule and explain it intuitively.
Why are the weights updated only when the prediction is incorrect?

## Write Answer 04:
**Perceptron Weights Update Rule:**

Wf=wi+eta*(y-y_hat)*xi

wi=Initial weight for xi

wf=Updated weight

eta=Learning rate

y=Actual label

y_hat= Predicted label

**Intuitive Explanation:**
If the prediction is correct then point is already on the correct side of the line sothat no need to move the line.

If the prediction is incorrect then point is not on the correct side of the line.Shift the line  sothat  this point moves toword the correct side.

 The weights updated only when the prediction is incorrect because when y_hat = y,no need to change the weight.If we change the weight after a correct prediction, it could damage the decision boundary.



# Question 05: [ Marks 10 ]

Why does a single-layer perceptron fail to solve the XOR problem?
What does this limitation tell us about the need for multilayer neural networks?

## Write Answer 05:
A single-layer perceptron can learn only linearly separable problems, meaning problems where classes can be separated by a single straight line.
The XOR problem is not linearly separable.
Class 1 points are (0,1), (1,0) and
Class 0 points are (0,0), (1,1).
No single straight line can separate these two sets of points in 2D space.Since a single-layer perceptron uses a linear decision boundary: w₁x₁ + w₂x₂ + b=0

It cannot correctly classify XOR, no matter how the weights are chosen.

*What This Limitation Tells Us*

Single-layer perceptrons cannot model non-linear relationships.

A multilayer perceptron (MLP) with at least one hidden layer can solve XOR by:

1.Learning multiple linear boundaries in hidden neurons

2.Combining them to form a non-linear decision boundary

# Question 07: [ Marks 20 ]

What are the main limitations of a single-layer perceptron?
How does the idea of using more than one layer (MLP) help overcome these limitations at a high level accoriding to you?



## Write Answer 07:

**Main Limitations of a Single-Layer Perceptron**

1. Can only learn linearly separable problems. A single-layer perceptron creates one linear decision boundary.It fails when classes cannot be separated by a straight line.


2. It cannot model non-linear relationships.For real-world data is often non-linear.A single perceptron has no mechanism to bend or combine boundaries.

3. Very Limited Representation Capacity.It acts like a simple threshold function.IT cannot learn hierarchical or abstract features.

4. No Hidden Layers for Feature Transformation.

**Multilayer Perceptrons (MLPs) Overcome These Limitations**

1. A hidden layers enable non-linear decision boundaries.Combining many such neurons produces non-linear shapes.

2. The activation Functions Add Non-Linearity.Hidden layers use non-linear activations (ReLU, sigmoid, tanh,Leaky ReLU).

3. Hierarchical Feature Learning

4. Provide universal approximation power.An MLP with at least one hidden layer can approximate any continuous function.
