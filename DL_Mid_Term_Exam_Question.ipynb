# Question 04: [ Marks 15 ]

Create a scalar tensor x = 3.0 with gradient tracking enabled. <br>
Compute y = x² + 4x. <br>
Perform backpropagation. <br>
Print the gradient of x. <br>

Write the full PyTorch code yourself. <br>

## Write Answer 04:



```python
# 1. Create scalar tensor with gradient tracking
x = torch.tensor(3.0, requires_grad=True)
# 2. Compute y = x^2 + 4x
y = x**2 + 4*x
# 3. Backpropagation
y.backward()
# 4. Print gradient of x
print("Value of x:", x.item())
print("Value of y:", y.item())
print("Gradient dy/dx:", x.grad.item())
```

# Question 05: [ Marks 10 ]

Create two tensors:
w = 2.0 (requires_grad=True) <br>
b = 1.0 (requires_grad=True) <br>

Let input x = 3.0. <br>
Compute z = w*x + b. <br>
Use Sigmoid to get y_pred.
 <br>

Define Binary Cross Entropy manually (without built-in loss). <br>
Compute gradients using backward(). <br>
Print gradients of w and b. <br>

Write full code.

## Write Answer 05:

```python
# 1. Create parameters with gradient tracking
w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

# 2. Input and target
x = torch.tensor(3.0)
y = torch.tensor(1.0)

# 3. Linear combination
z = w * x + b

# 4. Sigmoid activation
y_pred = torch.sigmoid(z)

# 5. Manual Binary Cross Entropy
loss = -(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred))

# 6. Backpropagation
loss.backward()

# 7. Print results
print("z:", z.item())
print("y_pred:", y_pred.item())
print("Loss:", loss.item())
print("Gradient dw:", w.grad.item())
print("Gradient db:", b.grad.item())
```

```python
# 1. Create tensor with gradient tracking
x = torch.tensor(2.0, requires_grad=True)

# 2. Forward computations
y1 = x**2
y2 = 3 * x
L = y1 + y2

# 3. Backpropagation
L.backward()

# 4. Print gradient
print("Value of x:", x.item())
print("Value of L:", L.item())
print("Gradient dL/dx:", x.grad.item())
```
