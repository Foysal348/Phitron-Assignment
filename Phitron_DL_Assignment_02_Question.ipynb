# Question 01: [ Marks 10 ]
Explain gradient descent in detail. What is the main idea behind it? How does it help in minimizing the loss function? Explain the role of learning rate and describe what happens if the learning rate is too high or too low.

**Write** Answer 01:

Gradient Descent is an iterative optimization algorithm in neural networks that minimizes the loss function by adjusting parameters, such as biases and weights, in the direction opposite the gradient.
Its main purpose is to find the best parameters that produce minimal loss.

**Main Idea Behind Gradient Descent**

Imaging you are standing on a mountain and want to reach the lowest point(valley).Suppose,
The height of the mountain refers to the loss function value.
Your position refers to the model parameters.
The slope direction tells you where to move to go downhill.
Gradient Descent works by:
1. Calculating the gradient of the loss function.
2. Moving in the opposite direction of the gradient.

Mathematically:

                              W = W - η.dL/d(W)
Where:

W = weights (parameters)

η = learning rate

dL/d(W) = gradient

So we:

1.Compute predictions (forward propagation)

2.Compute loss

3.Compute gradients (backpropagation)

4.Update weights

5.Repeat

**How It Minimizes the Loss Function**

At each step:

If gradient is positive → weight decreases

If gradient is negative → weight increases

If gradient is zero → we are at a minimum (or flat region)
Each update moves the parameters slightly closer to a point where the loss is smaller.

Adjust weights to reduce loss:

                              W = W - η.dL/d(W)

That's optimization.

**The Role of Learning Rate**

The learning rate is a hyperparameter that determines how big a step we take while updating the weights during gradient descent.

For this equation:

                              W = W - η.dL/d(W)

Where η refers to the learning rate.Learning rate tells us how far to move.

**If Learning Rate Is Too High**

1.Overshooting the minimal

2.Oscillation

3.Divergence (loss increases instead decreases)

Mathematically:

1.Updates become unstable

2.Loss graph zig-zags wildly

3.Training may never converge

**If Learning Rate Is Too Low**

Problems:

1.Very slow training

2.Gets stuck for a long time

3.May get trapped in local minima

# Question 02: [ Marks 10 ]

What is forward propagation in an artificial neural network? Explain step by step how data moves from the input layer to the output layer.

## Write Answer 02:

**Forward propagation in an artificial neural network**

Forward propagation is the process where input data moves through the neural network layer by layer from the input layer to the output layer to produce a final prediction or output.

It is called “forward” because information flows in the forward direction only (no weight updates yet).

**Step by step how data moves from the input layer to the output layer**

**Step 1: Input Layer (Receive Data)**

i. The network receives input features.

ii. Each input node represents one feature.

Example:

                                  X=[x1​,x2​,x3​,...,xn​]

The input layer does not perform computation; it just passes values forward.

**Step 2: Linear Transformation (Weighted Sum)**

Each neuron calculates a weighted sum:

                                        z = W⋅X + b

Where:

W = weights

X = input vector

b = bias

z = linear combination.

This step determines how strongly each input influences the neuron.

**Step 3: Apply Activation Function**

The linear output z is passed through an activation function:

                                              a = f(z)

f(.) demostrate the activation function such as ReLU, Leaky ReLU,Sigmoid, Tanh, etc.

**Step 4: Pass Output to Next Layer**

The activation output becomes the input for the next layer.

Process repeats:

                                        z^(l)=W^(l).a^(l-1)+b^(l)


Depending on task:

Sigmoid -> binary classification

Softmax -> multi-class classification

Linear activation -> regression.

# Question 03: [ Marks 15 ]
Why do we need activation functions in neural networks? What would happen if we only used linear transformations without any activation function?

## Write Answer 03:

**Why do we need activation functions in neural networks?**

Activation functions are essential in neural networks because they introduce non-linearity, enable learning complex parameters and control how signals propagate through the network. Without them, deep networks would behave like simple linear models.
A neuron computes:

                                              z = W.x + b

and then applies an activation:

                                                 a = f(z)

where,
f(⋅) is the activation function (ReLU, sigmoid, tanh,softmax,leaky ReLU, etc.).

**What happens if we only use linear transformations?**

Assume a network with multiple layers but no activation functions.

Each neuron typically computes:

                                                z = w^T.x + b
This is a linear transformation.
If we stack multiple layers without applying an activation function:

                                              Y = W3(W2( W1.X)
Simplified form

                                                  Y = W'.X
which is still just one linear transformation.

Activation functions (ReLU, sigmoid, tanh,leaky ReLU, etc.) apply a nonlinear mapping:

                                                  a=f(z)

This allows the network to approximate complex functions.


With activation functions:

1. Each layer learns a new representation.

2. Early layers detect simple patterns.

3. Later layers combine them into higher-level abstractions.

So,without activation functions:

1. Multiple layers add no extra expressive power.

2. Cannot learn non-linear relationships.

3. No hierarchical feature learning.

# Question 04: [ Marks 15 ]

Explain the sigmoid activation function. Where is it used? What are its advantages and disadvantages? What problem does it cause during training?

## Write Answer 04:

**Sigmoid Activation Function**

The sigmoid (logistic) activation function maps any real-valued input into a range between 0 and 1.

Mathematical form:

                                    σ(z)=1/1+e^(-z)

Where:

z = W.x + b (linear combination of inputs)

Output range: 0 < σ(z) < 1

Outputs can be interpreted as probabilities and importance for backpropagation in neural networks.

**Where is it used?**

*1. Binary Classification in Neural Networks(Output Layer)*

It uses the final layer of binary classifiers. It converts output into probability.
Example:
i. Email spam or not spam
ii. Disease or not disease

*2. Logistic Regression*

Sigmoid is the core function used to map linear outputs to probability values.

*3. Early Neural Networks in hidden layers*

Previously used in hidden layers, but now less common because of training issues.

**Advantages**

i. Outputs are bounded (0 to 1)

ii. Smooth and differentiable

iii. Simple mathematical form

**Disadvantages**

i. Not zero-centric

ii. Computationally havier than ReLU

iii. Saturation at extreme values

**Major problem during training**
The biggest issue with sigmoid is the vanishing gradient problem.
Derivative of sigmoid:

                              σ'(z) = σ(z)(1-σ(z))
Maximum derivative is only 0.25.

When output approaches 0 or 1 ,then derivative ≈ 0.

*During backpropagation:*

Gradients shrink as they pass through many layers.

Early layers receive very tiny updates.

*Consequences:*

1. Slow learning
2. Difficulty training deep networks
3. Poor convergence.

# Question 05: [ Marks 10 ]
Explain the softmax activation function in simple words. Why is it used in multi-class classification problems? How is it different from sigmoid?

## Write Answer 05:

**Softmax Activation Function**



The softmax activation function converts a list of numbers (logits) into probabilities that add up to 1.

Mathematical form

For outputs z1​,z2​,...,zn:

Screenshot 2026-02-18 232453.png

Suppose a model predicts animal classes:

Raw outputs (logits):

Cat = 2.0

Dog = 1.0

Bird = 0.1

After softmax:

Cat = 0.65

Dog = 0.24

Bird = 0.11

Model predicts Cat because it has highest probability.

**Why Softmax is Used in Multi-Class Classification**

Multi-class classification means:

One input belongs to one class out of many.

Examples:

i. Digit recognition (0–9)

ii. Animal classification

iii. Language identification

Softmax helps because:

i. Converts outputs into probabilities.

ii. Ensures probabilities sum to 1.

iii. Creates competition between classes

Works well with cross-entropy loss.

**Softmax vs Sigmoid**

**Sigmoid**

Sigmoid is an activation function that converts a single value into a probability between 0 and 1. Each output neuron works independently. It is commonly used for:

Binary classification (yes/no)

Multi-label problems where multiple classes can be true at the same time.

Example: an image may contain both a cat and a dog, each class probability is calculated independently.


**Softmax**

Softmax is an activation function used when we have multiple classes and the input belongs to only one class among them. It takes several output values (logits) and converts them into probabilities that add up to 1. Because of this normalization, all classes compete with each other — increasing the probability of one class automatically decreases the probabilities of others.

Example: digit recognition (0-9), where an image represents only one digit.

**Main Differences**

Softmax considers all outputs together and produces a probability distribution whose total sum is 1. It assumes classes are mutually exclusive.

Sigmoid treats each output separately, so probabilities do not need to sum to 1. Multiple classes can have high probabilities simultaneously.

Softmax is mainly used for multi-class (single-label) classification, while sigmoid is used for binary classification or multi-label classification.


# Question 06: [ Marks 20 ]

Explain the structure of an Artificial Neural Network in detail. Describe the role of the input layer, hidden layers, and output layer. How do weights, biases, and activation functions work together to transform input data into predictions? Illustrate your explanation with a simple example.

## Write Answer 06:

**Structure of an Artificial Neural Network (ANN)**

An Artificial Neural Network (ANN) is a computational model inspired by the human brain. It consists of interconnected layers of neurons that process input data step by step to produce predictions.

The basic structure includes:

i. Input Layer

ii. Hidden Layer(s)

iii. Output Layer

Each layer contains neurons connected through weights, biases, and activation functions.

**1. Input Layer**

The input layer is the first layer of the network.

*Role:*

i. Receives raw input data (features).

ii. Each neuron corresponds to one input feature.

It does not perform complex computations.It simply passes data forward.

**2. Hidden Layers**

Hidden layers are the intermediate layers between input and output.

*Role:*

i. Perform computations and feature extraction.

ii. Learn complex patterns by combining inputs.

iii. Each neuron calculates:

                                 z = W.x + b
and then applies an activation:

                                                 a = f(z)

where,
f(⋅) is the activation function (ReLU, sigmoid, tanh,softmax,leaky ReLU, etc.).

**3. Output Layer**

The output layer produces the final prediction.

Role:

i. Converts processed information into final output.

Examples:

Sigmoid activation -> binary classification.

Softmax -> multi-class classification.

**How Weights, Biases, and Activation Functions Work Together**

(1) Weights

Control the importance of each input.

Larger weight → stronger influence.

Mathematically:

                                   z=w1.​x1​+ w2.​x2​+⋯+wn.​xn​

(2) Bias

Adds flexibility by shifting the activation.

Helps model fit data even when input is zero.

                                       z = W.X +b

(3) Activation Function

Applies non-linear transformation:

                                        a = f(z)

where,
f(⋅) is the activation function (ReLU, sigmoid, tanh,softmax,leaky ReLU, etc.).

**Step-by-Step Transformation (Forward Flow)**

1. Input features enter network.

2. Multiply inputs by weights.

3. Add bias.

4. Apply activation function.

5. Pass result to next layer.

6. Repeat until output layer produces prediction.

**Simple Example**

Suppose we want to classify whether a student passes an exam.

Inputs:

i. Study hours

ii. Attendance
